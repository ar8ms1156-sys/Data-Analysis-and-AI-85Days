```python
# داتا مبيعات وجبة "كبسة الدجاج" لمطعم في الرياض خلال الأسبوع الماضي
sales_data = [
    {"day": "الأحد", "sales_sar": 1200, "meals_sold": 40},
    {"day": "الإثنين", "sales_sar": 1350, "meals_sold": 45},
    {"day": "الثلاثاء", "sales_sar": 1100, "meals_sold": 37},
    {"day": "الأربعاء", "sales_sar": 1500, "meals_sold": 50},
    {"day": "الخميس", "sales_sar": 2100, "meals_sold": 70},
    {"day": "الجمعة", "sales_sar": 2550, "meals_sold": 85},
    {"day": "السبت", "sales_sar": 1800, "meals_sold": 60}
]
```

```python
# حساب إجمالي عدد الوجبات المباعة خلال الأسبوع
total_meals = sum(day["meals_sold"] for day in sales_data)

# عرض النتيجة
print("إجمالي الوجبات المباعة في الأسبوع:", total_meals)
```

```python
# حساب متوسط عدد الوجبات المباعة في اليوم
avg_daily_meals = total_meals / len(sales_data)

# عرض النتيجة
print("متوسط المبيعات اليومية للوجبات:", avg_daily_meals)
```

```python
import math

# إضافة هامش أمان 10% على المتوسط لحماية المطعم من نفاد الوجبات
recommended_prep = avg_daily_meals * 1.10

# تقريب الرقم لأعلى لأقرب رقم صحيح (لأن الوجبات أرقام صحيحة)
final_prepared_meals = math.ceil(recommended_prep)

# عرض التقرير النهائي للمطعم
print("الكمية الموصى بتحضيرها يومياً لتفادي الهدر:", final_prepared_meals, "وجبة")
```
